```python
import pandas as pd
import matplotlib.pyplot as plt
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y')
plt.figure(figsize=(12,6))

plt.plot(df['Date'], df['Close'])

plt.xlabel('Date')
plt.ylabel('Closing Price ($)')
plt.title('Tesla Stock Closing Price Over Time')

plt.show()
```

The above graph shows the increase in Telsa prices over the years.
This is important as it quantafies the return on investment one could expect from Tesla stock.
Additionally this type of graph is essential to show changes is many phenomenom such as air quality.


```python
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y')
plt.figure(figsize=(12,6))

plt.plot(df['Date'], df['Volume'])

plt.xlabel('Date')
plt.ylabel('Stock Volume')
plt.title('Tesla Stock Volume Over Time')

plt.show()
```

The Tesla stock volume seems to change generally consistently over time since 2014 when the volume initially spiked high, with another big spike in the early 2020s and then a significant decrease towards 2022.
